```python
import pandas as pd
#Importando um arquivo .csv dos dados de temperatura dos municípios
df = pd.read_csv('https://raw.githubusercontent.com/andr3nf/clima_municipios_brasileiros/refs/heads/main/clima_historico_municipal.csv')
```

```python
cidade = df.loc[df['nome_municipio'] == 'Santa Rosa']

cidade
```

```python
deficit = cidade[[
    'def_jan', 'def_fev', 'def_mar', 'def_abr',
    'def_mai', 'def_jun', 'def_jul', 'def_ago',
    'def_set', 'def_out', 'def_nov', 'def_dez'
]]

mesMaiorDeficit = deficit.idxmax(axis=1)

mesMaiorDeficit
```

```python
mediaVerao = cidade[
    ['temp_media_dez', 'temp_media_jan', 'temp_media_fev']
].mean(axis=1)

print(mediaVerao)
```

```python
mediaInverno = cidade[
    ['temp_media_jun', 'temp_media_jul', 'temp_media_ago']
].mean(axis=1)

print(mediaInverno)
```

```python
mediaPrimavera = cidade[
    ['temp_media_set', 'temp_media_out', 'temp_media_nov']
].mean(axis=1)

print(mediaPrimavera)
```

```python
mediaOutono = cidade[
    ['temp_media_mar', 'temp_media_abr', 'temp_media_mai']
].mean(axis=1)

print(mediaOutono)
```
