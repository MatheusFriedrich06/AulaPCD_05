```python
import json
import csv
import random

numerosAleatorios = random.sample(range(1, 101), 20)

valores = []
for x in numerosAleatorios:
    valores.append ({"valorOriginal": x,
                 "valorX2": x * 2,
                 "valorX10": x * 10 })

with open("valores.txt", "w") as f:
    f.write("valorOriginal\tvalorX2\tvalorX10\n")
    for v in valores:
        f.write(f"{v['valorOriginal']}\t{v['valorX2']}\t{v['valorX10']}\n")


with open("valores.json", "w") as f:
    json.dump(valores, f, indent=4)

with open("valores.csv", "w", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=["valorOriginal", "valorX2", "valorX10"])
    writer.writeheader()
    writer.writerows(valores)


print (numerosAleatorios)
print (valores)

```
